# Fungi Simulation in Crop Field.

This jupyter notebook simulates the way fungi behave in a crop field by using the random walkers algorithm.
It is also able to generate a SIR model side by side to the simulation for the succeptible, infected and recovered plants.

### Abstract

This project was aimed to simulate the behaviour of a fungi infection in a crop field, though it can also be used to simulate other growth behaviours, like the behaviour of viruses, bacteria, growth in animal territory and etc, by adjusting the simulation parameters.

Firstly, the simulation happens in a grid of points and tries to replicate the infection fenomena by using the random walkers algorithm. Starting with a certain number of infected points, in each step all infected points capable of infecting neighbors select a random neighbor to infect using the random walkers algorithm. There is also a chance for the infected points to be recovered or killed by the fungi. When they recover, they no longer are infected - so they can not infect other points - and when they are killed, they become permanently infected - so they can not be recovered. The code also has functions to generate walls or obstacles in the crop field and all functions are heavily documented - so it should be easy to use those at will to see how the infection will change with the addition of walls, obstacles and etc.

The need for a simulation comes from the difficult equation solution when considering that fungi will still infect other plants even when the plant is killed. When there's no death or the death behaves differently, it is much easier to predict the behaviour of the infected population. In the theoric basis it was explained what the expected recover parameter should be in order for the infection to be contained and how to achieve the *difficult* equation that shows the need for such a simulation.

To finish, when testing with the parameters of common fungi, the fungi was behaving as expected by those shown in real life, also showing some fenomena which correspond to real life fungi. Therefore, it can be concluded that the simulation can show an estimate of the behaviour of real-life infection fenomena.

## Theoric Basis

### SIR Model

The code was written in a way that the number of points that *already* were infected $T$ - therefore, all points excluding the succeptible; of recovered points $R$ and dead points $M$ were saved. Below is the equation for $T_n$, where $n$ is the number of the simulation step:

$$T_n = I^*_n + R_n + M_n$$

Where $I^*_n$ is the number of exclusive infected points - red points of the simulation, therefore excluding dead points; $R_n$ are the recovered points and $M_n$ are the dead points.

$I^*$ can be represented as followed:

$$I^*_n = I_{n} - M_{n}$$

$I_n$ is the number of points that can infect, including dead points.

#### Types of points in the code:

The number of succeptible points $S_n$ is the total number of plants (which is the area of the field A) minus the number of points that were already infected:

$$S_n = A - T_{n}$$

The number of exclusive infected points is derived from the first equation:

$$I^*_n = T_n - R_n - M_n$$

The number of recovered points is already saved as $R_n$.

### Theoric Model

#### Function for infected, dead, recovered and suceptible.

In each simulation step $n$ a random neighbor of a infected point is chosen to be infected with a $\beta$ probability. The chance $p(v)$ of this point not being already infected depends on the number of neighbors infected $v$ of this point:

$$p(v) = \beta \left(1 - \frac{v}{4}\right)$$

The number of infected points $N_v$ that contain a number of infected neighbors $v$ can be described as a percentage $P_v$ times the total number of infected points $I$:

$$N_v = IP_v$$

Knowing that the infection in a point already infected doesn't grow the number of infected points, the number of infected points $N_{v_n}$ obtained by the fungi, considering $v$ can be calculated with the probability $p(v)$ of these point infecting a safe point:

$$N_{v_n} = p(v)N_{v_{n-1}} = p(v)I_{n-1}P_{v_{n-1}}$$

Considering all possibilities of neighbors:

$$N_{0_n} = \frac{4}{4} \beta P_{0_{n-1}}I_{{n-1}}$$
$$N_{1_n} = \frac{3}{4} \beta P_{1_{n-1}}I_{{n-1}}$$
$$N_{2_n} = \frac{2}{4} \beta P_{2_{n-1}}I_{{n-1}}$$
$$N_{3_n} = \frac{1}{4} \beta P_{3_{n-1}}I_{{n-1}}$$
$$N_{4_n} = \frac{0}{4} \beta P_{4_{n-1}}I_{{n-1}}$$

The total number of infected points $N_n$ obtained is:

$$N_{0_n} + N_{1_n} + N_{2_n} + N_{3_n} + 0 \equiv N_n$$

Also, the sum of all the percentages should be equal to 1:

$$P_{0_{n-1}} + P_{1_{n-1}} + P_{2_{n-1}} + P_{3_{n-1}} + P_{4_{n-1}} = 1$$

Therefore:

$$N_n = \beta P_{0_{n-1}}I_{{n-1}} + \frac{3}{4} \beta P_{1_{n-1}}I_{{n-1}} + \frac{2}{4} \beta P_{2_{n-1}}I_{{n-1}} + \frac{1}{4} \beta P_{3_{n-1}}I_{{n-1}}$$

The number of infected plants $I_n$ is not exactly the sum of itself with $N_n$ because there is another process that lowers the number of exclusively infected in each simulation step $n$.

This process is the recover of the plants, which lowers the amount of exclusive infected points with probability $\gamma$ and plants in this new state cannot die (therefore, permanently becoming infected). There is also dead plants - which can die with probability $\mu$
when infected - that still *infect* other points and can not be recovered.

The number of recovered points $R_n$ is:

$$R_n = R_{n-1} + \gamma I^*_{n-1}$$

The number of dead points $M_n$ is:

$$M_n = M_{n-1} + \mu I^*_{n-1}$$

The rate of change of infected points is:

$$\Delta I = N_{n-1} - \Delta R$$

Then: 

$$I_n = I_{n-1} + N_{n-1} - \Delta R$$

The number of infected points $I_n$ is:

$$I_n = I_{n-1} + N_{n-1} - \gamma I^*_{n-1}$$

Changing $I$ with $I^* + M$:

$$I^*_n + M_n = I^*_{n-1} + M_{n-1} + N_{n-1} - \gamma I^*_{n-1}$$
$$I^*_n = I^*_{n-1} + N_{n-1} - \gamma I^*_{n-1} - \Delta M $$

With the definition of $M_n$:

$$\Delta M = \mu I^*_{n-1}$$

Then:

$$I^*_n = I^*_{n-1} + N_{n-1} - \gamma I^*_{n-1} - \mu I^*_{n-1} $$

Expanding $N_{n-1}$:

$$I^*_n = I^*_{n-1} + \beta P_{0_{n-1}}I_{{n-1}} + \frac{3}{4} \beta P_{1_{n-1}}I_{{n-1}} + \frac{2}{4} \beta P_{2_{n-1}}I_{{n-1}} + \frac{1}{4} \beta P_{3_{n-1}}I_{{n-1}} - \gamma I^*_{n-1} - \mu I^*_n $$
$$I^*_n = I_{n-1}\left[\beta P_{0_{n-1}} + \frac{3}{4} \beta P_{1_{n-1}} + \frac{2}{4} \beta P_{2_{n-1}} + \frac{1}{4} \beta P_{3_{n-1}} \right] + I^*_{n-1} (1 - \gamma - \mu) $$

Changing $I$ with $I^* + M$ again:

$$I^*_n = [I^*_{n-1} + M_{n-1}]\left[\frac{\beta}{4}(4P_{0_{n-1}} + 3 P_{1_{n-1}} + 2 P_{2_{n-1}} + P_{3_{n-1}}) \right] + I^*_{n-1} (1 - \gamma - \mu)$$

It is possible to rename the following as a percentage parameter $P_{n-1}$:

$$4P_{0_{n-1}} + 3 P_{1_{n-1}} + 2 P_{2_{n-1}} + P_{3_{n-1}} \equiv P_{n-1}$$

Finally:

$$I^*_n = I^*_{n-1}\left[\frac{\beta}{4}P_{n-1} + 1 - \gamma - \mu \right] + M_{n-1}\frac{\beta}{4}P_{n-1}$$

Finding a solution of the above expression is extremely difficult since $M$ is written in terms of $I^*_n$ and vice-versa.

**Solution with no death rate:**

When $\mu = 0$:

$$I^*_n = I^*_{n-1}\left[\frac{\beta}{4}P_{n-1} + 1 - \gamma \right]$$

Also, the number of exclusive infected isequal to the total number of infected, because $M_n = 0$:

$$I_n = I^*_n + M_n$$
$$I_n = I^*_n$$

Solving $I_n$:
$$I_n = I_0 \prod_0^n\left(1 - \gamma + \frac{\beta}{4}P_{n-1} \right)$$

It is noticeable the exponential grow in the number of infected in the above function.

*The script implemented variated probabilities inside a margin of values. In that case, it is used an average probability.
Ex.: $\bar{\gamma}, \bar{\mu}, \bar{\beta}$*

#### Evaluating convergence of fungi: Total infection or total death.

It is necessary to evaluate the values of $P_{n-1}$:

$$P_{n-1} = 4P_{0_{n-1}} + 3 P_{1_{n-1}} + 2 P_{2_{n-1}} + P_{3_{n-1}}$$

The only moment where there is a infected point with no infected neighbors is the initial step of the simulation *when it starts with only 1 point*. This way:

$$P_{n} = 3 P_{1_{n}} + 2 P_{2_{n}} + P_{3_{n}} \text{, }  n \ge 1$$

But:
$$P_{0_{n-1}} + P_{1_{n-1}} + P_{2_{n-1}} + P_{3_{n-1}} + P_{4_{n-1}} = 1$$

It is possible to achieve the extreme values of $P_{n}$:

$$P_n = 3 \text{, when } P_{1_n} = 1 \text{, for } n \ge 1$$
$$P_n = 0 \text{, when } P_{4_n} = 1 \text{, for } n \ge 1$$

Finally, a interval of $P_n$ is obtained for $n \ge 1$ or $I_0 > 1$:
$$P_n \in [0, 3], \hspace{1em} n \ge 1 \text{ or } I_0 > 1$$
$$P_n = 4 \text{ when }  n = 0 \text{ and } I_0 = 0$$

The equation obtained of $I_n$ demonstrates the exponential growth of the infected population. It is possible to evaluate when the population will get totally infected or if the recover parameter will overcome the death rate:

$$\lim_{n \to \infty} I_n = \lim_{n \to \infty} I_0 \prod_0^n\left(1 - \gamma + \frac{\beta}{4}P_{n-1} \right)$$

When $n \rightarrow \infty$, $(n - 1) \rightarrow n$:

$$\lim_{n \to \infty} I_n = I_0 \lim_{n \to \infty}  \prod_0^n\left(1 - \gamma + \frac{\beta}{4}P_{n} \right)$$

First case:

$$1 - \gamma + \frac{\beta}{4}P_{n} > 1 $$
$$\lim_{n \to \infty} I_n = \infty$$

This case can be interpreted as an infinite growth of the infected population. It's a total domination of the fungi - so most of the crop field will become infected.

Second case:

$$1 - \gamma + \frac{\beta}{4}P_{n} < 1$$
$$\lim_{n \to \infty} I_n = 0$$

The number of infected goes to $0$ as time passes. So the fungi will infect but eventually the recover of the plants will overcome the infection rate and the crops will survive.

Third case:

$$1 - \gamma + \frac{\beta}{4}P_{n} = 1$$
$$\lim_{n \to \infty} I_n = I_0$$

The number of infected is constant throughout the simulation. There will always be infected plants, but shortly after that they will be recovered, maintaining the number of infected mostly constant through time.

Evaluating $P_n$ outcomes:

$$1 - \gamma + \frac{\beta}{4}P_{n} > 1$$
$$\gamma < \beta\frac{P_n}{4}$$

The above inequation evaluates the first case, where the infection dominates the crops.

For $P_n = 0$:

$$\gamma < 0$$

The recover parameter can not be negative, therefore it is impossible for the fungi to grow indefinitely when $P_n = 0$.

For $P_n = 3$:

$$\gamma < \beta\frac{3}{4}$$

Since $3$ is the maximum value for $P_n$, with $\gamma > \frac{3}{4} \beta$ it is confirmed that the infection will lose.

In reality, the number $P_n$, when $n \ge 1$ will start as $3$ and lower to a small number as time passes because of the growth of points that are inside the "*infection circle*" - which contain $4$ neighbors. So the percentage of points with 4 neighbors $P_4$ will grow with time much faster than points with 1, 2 or 3 neighbors, but the contribution from these points will always exist.

## Utils

```python
# Importing modules
%matplotlib tk
from matplotlib.animation import FuncAnimation
from functools import partial

#from IPython.display import HTML

import matplotlib.pyplot as plt
import numpy as np
```

```python
# Useful 

# The 4 directions of the random walk
directions = [np.asarray([0, 1]), np.asarray([1, 0]), 
              np.asarray([0, -1]), np.asarray([-1, 0])] 

# Default seed
seed = 42

# Generates a pseudo-random number
def Random():
    global seed;
    a, b, c = 1029, 221591, 1048576;
    seed = (seed*a + b) % c;
    return float(seed) / c;

# Returns a random number between start and end
def RandomRange(start, end):
    return (end - start) * Random() + start
```

### Walker Function

```python
# One walker step from random walker algorithm
def WalkerStep(point):
    randomIdx = int(4*Random())

    return point + directions[randomIdx]
```

### Geometry Functions

```python
# Generates integer points given a function, start and end.
def Wall(f, start, end):
    xVals = np.arange(int(start), int(end) + 1, 1)
    yVals = f(xVals).astype(int)

    return [tuple(k) for k in np.column_stack((xVals, yVals))]
```

```python
# Creates field border
def Rectangle(center, width, height):
    # Calculating halfs 
    halfWidth  = int(width / 2)
    halfHeight = int(height / 2)
    
    points = []
    for i in [-1, 1]:
        def f(x): 
            correction = -1 if width % 2 == 0 else 0
            if i == -1: correction = 0
                
            return 0*x + center[0] + halfWidth * i + correction
        def g(x): 
            correction = -1 if height % 2 == 0 else 0
            if i == -1: correction = 0
                
            return 0*x + center[1] + halfHeight * i + correction

        # X bounds
        start = center[1] - halfHeight
        if width % 2 == 0:
            end = center[1] + halfHeight - 1
        else:
            end = center[1] + halfHeight
            
        xBound = Wall(f, start, end)
            
        # Y bounds
        start = center[0] - halfWidth
        if height % 2 == 0:
            end = center[0] + halfWidth - 1
        else: 
            end = center[0] + halfWidth 
            
        yBound = Wall(g, start, end)
        
        points.extend([(p[1], p[0]) for p in xBound])
        points.extend(yBound)
        
    return points
```

```python
# Checks if a points is inside the field 
def InsideRectangle(point, center, width, height):
    # Correction if value is odd or even
    xCorrection = -1 if width % 2 == 0 else 0
    yCorrection = -1 if width % 2 == 0 else 0
    
    xBound = center[0] + width  // 2
    yBound = center[1] + height // 2
    
    if point[0] > center[0] + xBound + xCorrection or \
       point[0] < center[0] - xBound or \
       point[1] > center[1] + yBound + yCorrection or \
       point[1] < center[1] - yBound:
        return False
    
    return True
```

```python
# Creates initial circle of infection
def InfectionCircle(radius):
    points = []
    radius_sq = radius ** 2
    for x in range(-radius, radius + 1):
        for y in range(-radius, radius + 1):
            if x**2 + y**2 > radius_sq: continue
                
            points.append((x, y))
    
    return points
```

### Animation Functions

```python
# Init function for FuncAnimation
def Init(artSimulation, artSIR):
    # Extracting previous artists
    recover, plague, death = artSimulation
    line_S, line_I, line_R = artSIR
    
    recover.set_data([], [])
    plague.set_data([], [])
    death.set_data([], [])
    
    line_S.set_data([], [])
    line_I.set_data([], [])
    line_R.set_data([], [])

    return [plague, recover, line_S, line_I, line_R]
```

```python
# Function that set artists for FuncAnimation
def SetArtists(frame, fieldArea, infoSimulation, artSimulation, infoSIR, artSIR):
    # Saving points to display on the simulation map (ax1)
    xRecover, yRecover = [], []
    xPlague, yPlague = [], []
    xDeath, yDeath = [], []
    for point, infection in infoSimulation.items():
        if infection.recovered:
            xRecover.append(point[0])
            yRecover.append(point[1])
            continue
            
        if infection.dead:
            xDeath.append(point[0])
            yDeath.append(point[1])
            continue
            
        xPlague.append(point[0])
        yPlague.append(point[1])

    # Extracting previous SIR information
    recovered, infected, dead = infoSIR

    time_steps = range(frame + 1) 
    susceptible_curve = fieldArea - infected[:frame+1]
    infected_curve = infected[:frame+1] - recovered[:frame+1] - dead[:frame+1]
    recovered_curve = recovered[:frame+1]

    # Extracting previous artists
    recover, plague, death = artSimulation
    line_S, line_I, line_R = artSIR

    # Saving points to display on SIR (ax2)
    recover.set_data(xRecover, yRecover)
    plague.set_data(xPlague, yPlague)
    death.set_data(xDeath, yDeath)
    
    # Atualiza as linhas do gráfico SIR
    line_S.set_data(time_steps, susceptible_curve)
    line_I.set_data(time_steps, infected_curve)
    line_R.set_data(time_steps, recovered_curve)
```

## Simulation

```python
# Point that represents a fungi in the field
class Point:
    def __init__(self, active=True):
        self.recovered = False
        self.dead = False
        
        self.active = active

    def Recover(self):
        self.recovered = True
        
    def Kill(self):
        self.dead = True
```

```python
# Simulation step
def SimulationStep(frame, infectionRange, recoverRange, deathRange, T,
                   fieldArea, infoSimulation, artSimulation, infoSIR, artSIR):
    # Extracting previous information
    recovered, infected, dead = infoSIR

    # Transforming to int for correct indexing
    frame = int(frame) 
        
    # Summing with previous number of recovered, infected and dead
    if frame >= 1:
        recovered[frame] = recovered[frame - 1]
        infected[frame] = infected[frame - 1]
        dead[frame] = dead[frame - 1]

    # Stop animating when time reaches the end
    if frame >= T - 1: infectionAnimation.pause()

    # Generating random chances for current step 
    b = RandomRange(infectionRange[0], infectionRange[1])
    g = RandomRange(recoverRange[0], recoverRange[1])
    m = RandomRange(deathRange[0], deathRange[1])
    
    curInfected = infoSimulation.copy()
    # Every infected point does a walker step
    for point, infection in curInfected.items():
        # Skipping recovered points or dead points
        if infection.recovered: continue
            
        if not infection.dead:
            # Chance of recovering plant
            if Random() < g:
                infoSimulation[point].Recover()
                recovered[frame] += 1
                continue
    
            # Chance of killing plant
            if Random() < m:
                infoSimulation[point].Kill()
                dead[frame] += 1
            
        # Skipping inactive points
        if not infection.active: continue

        # Unactivating points surrounded by others
        for curDir in directions:
            if tuple(np.asarray(point) + curDir) in infoSimulation:
                infoSimulation[point].active = False
                continue
                
            infoSimulation[point].active = True
            break
            
        # If point is not active or infection chance
        # was not achieved, skip.
        if not infoSimulation[point].active or Random() > b: continue
            
        newInfection = WalkerStep(np.asarray(point))
        if tuple(newInfection) in infoSimulation: continue
        
        infoSimulation[tuple(newInfection)] = Point()
        infected[frame] += 1
        
    SetArtists(frame, fieldArea, infoSimulation, artSimulation, infoSIR, artSIR)

    # Extracting artists to return
    recover, plague, death = artSimulation
    line_S, line_I, line_R = artSIR
    
    return [plague, recover, death, line_S, line_I, line_R]
```

The function below starts the simulation and it is where the functions to generate walls should be called, if desired.

```python
# Function that start the simulation and animates it
def Simulate(infectionRange, recoverRange, deathRange, T, fungiStart, fungiRadius, fieldCenter, width, height):
    # Calculating field area
    fieldArea = width * height

    # Creating initial infection circle and setting conditions
    allInfected = {}
    for i in InfectionCircle(fungiRadius):
        newPoint = tuple(np.asarray(i) + fungiStart)
        
        if not InsideRectangle(newPoint, fieldCenter, width, height):
            continue
            
        allInfected[newPoint] = Point()
    
    # Setting SIR arrays
    global recovered; recovered = np.zeros(T)
    global infected; infected = np.zeros(T)
    global dead; dead = np.zeros(T)
    
    # Initial count and infected values
    infected[0] = len(allInfected)
    
    # Adding walls
    # The following code adds the border of the crop field
    walls = Rectangle(fieldCenter, width + 2, height + 2)
    for point in walls:
        allInfected[point] = Point(False)
        allInfected[point].Kill()

    # Add more walls HERE if it is desired to see how the fungi behaviour changes 
    # with the addition of obstacles in the field.
    
    # The following commented code creates a wall with a small gap in the top.
    # def h(x): return 0 * x
    # line = Wall(h, -height/2, height/2) 
    # for i, point in enumerate(line):
    #     if i > len(line) - 5: continue
    #     newP = (point[1], point[0])
    #     allInfected[newP] = Point(False)
    #     allInfected[newP].Kill()
    
    # Plotting
    
    # Creates the figure and two subplots side by side
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))
    
    # --- First graph configuration: Fungi infection simulation ---
    ax1.set_xlim(-width/2 - 5, width/2 + 5)
    ax1.set_ylim(-height/2 - 5, height/2 + 5)
    ax1.set_title("Simulação da Infecção por Fungo")
    ax1.set_xlabel("Posição X")
    ax1.set_ylabel("Posição Y")
    ax1.set_aspect('equal')
    ax1.grid()
    
    # Correction if value is odd or even
    xCorrection = 0 if width % 2 == 0 else 1
    yCorrection = 0 if width % 2 == 0 else 1
    
    # Generating field points
    xVals = range(fieldCenter[0] - width//2, fieldCenter[0] + width//2 + xCorrection)
    yVals = range(fieldCenter[1] - height//2, fieldCenter[1] + height//2 + yCorrection)
    
    fieldPoints = []
    for x in xVals:
        for y in yVals:
            fieldPoints.append((x, y))
    fieldPoints = np.asarray(fieldPoints)
    
    # Plots for fixed points
    susceptible, = ax1.plot(fieldPoints[:, 0], fieldPoints[:, 1], marker='.', color='khaki', linestyle='', label='Suscetível')
    
    # Plots for points that change
    recover, = ax1.plot([], [], marker='.', color='seagreen', linestyle='', label='Recuperado')
    plague, = ax1.plot([], [], marker='.', color='indianred', linestyle='', label='Infectado')
    death, = ax1.plot([],[], marker='.', color='k', linestyle='', label="Morto")
    
    ax1.legend()
    
    # --- Second graph configuration: SIR Model ---
    ax2.set_xlim(0, T)
    ax2.set_ylim(0, fieldArea * 1.05) # Um pouco de espaço no topo
    ax2.set_title("Evolução do Modelo SIR")
    ax2.set_xlabel("Tempo (dias)")
    ax2.set_ylabel("Número de Indivíduos")
    ax2.grid()
    
    # SIR line curves
    line_S, = ax2.plot([], [], label="Suscetível", c="navy")
    line_I, = ax2.plot([], [], label="Infectado", c="crimson")
    line_R, = ax2.plot([], [], label="Recuperado", c="Teal")
    ax2.legend()
    
    # Better layout
    fig.tight_layout()
    
    # --- Animating ---
    artSIR = (line_S, line_I, line_R)
    artSimulation = (recover, plague, death)
    infoSIR = (recovered, infected, dead)
    
    global infectionAnimation
    infectionAnimation = FuncAnimation(fig, 
                                       partial(SimulationStep, infectionRange=infectionRange, recoverRange=recoverRange,
                                               deathRange=deathRange, T=T, fieldArea=fieldArea,
                                               infoSimulation=allInfected,  artSimulation=artSimulation,  
                                               infoSIR=infoSIR, artSIR=artSIR), 
                                       init_func=partial(Init,  artSimulation=artSimulation, 
                                                         artSIR=artSIR), 
                                       interval=1, frames=T, blit=True)
    plt.show()
    # Fallback tk not functional
    #HTML(infectionAnimation.to_html5_video())
```
